```python
hoTen=input("Nhập tên học sinh:")
toan=float(input("Nhập điểm toán:"))
van=float(input("Nhập điểm Văn:"))
tin=float(input("Nhập điểm tin:"))
dtb=(toan+van+tin)/3
if dtb<5:
    xeploai="Yếu"
elif dtb<7:
    xeploai="Trung bình"
elif dtb<8:
    xeploai="Khá"
else:
    xeploai="Giỏi"
print("ĐTB của học sinh "+ hoTen +" là:" +str(dtb)+ " Xếp loại:"+str(xeploai))
```
