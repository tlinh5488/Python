```python
import math
def tamgiac(a, b, c):
    if a+b <= c or a+c <= b or b+c <= a :
        return "Không phải là tam giác"
    a,b,c = sorted([a,b,c])
    if a==b==c:
        loai= "Tam giác đều"
    elif a==b and a*a + b*b == c*c:
        loai= "Tam giác vuông cân"
    elif a==b or b==c:
        loai= "Tam giác cân"
    elif a*a + b*b == c*c:
        loai= "Tam giác vuông"
    else:
        loai="Tam giác thường"
    p= (a+b+c)/2
    S= math.sqrt(p*(p-a)*(p-b)*(p-c))
    return loai, S
a=float(input("Nhập độ dài cạnh a:"))
b=float(input("Nhập độ dài cạnh b:"))
c=float(input("Nhập độ dài cạnh c:"))
loai, S = tamgiac(a, b, c)
print(str(loai) + " Diện tích là: " + str(S))
```

```python
def tienDien(soLuong):
    if soLuong <= 100:
        tienDien= soLuong*1418
    elif soLuong <=150:
        tienDien= 100*1418+(soLuong-100)*1622
    elif soLuong <=200:
        tienDien=100*1418+50*1622+(soLuong-150)*2044
    elif soLuong <=300:
        tienDien=100*1418+50*1622+100*2044+(soLuong-200)*2210
    elif soLuong <=400:
        tienDien=100*1418+50*1622+100*2044+100*2210+(soLuong-300)*2361
    else:
        tienDien=100*1418+50*1622+100*2044+100*2210+100*2361+(soLuong-400)*2420
    return tienDien
chisoTruoc=int(input("Nhập chỉ số tháng trước:"))
chisoSau=int(input("Nhập chỉ số tháng sau:"))
soLuong=chisoSau - chisoTruoc
tiendien = tienDien(soLuong)
print("Tiền điện của bạn là: " + str(tiendien))
```

```python
def sothang(tiengui, tiennhan, lai):
    thang=0
    while tiengui<tiennhan:
        tiengui=tiengui + tiengui *lai /100
        thang +=1
    return thang
tiengui=float(input("Số tiền đã gửi là: "))
tiennhan=float(input("Số tiền đã nhận là: "))
lai=float(input("Nhập lãi suất 1 tháng: "))
thang=sothang(tiengui, tiennhan, lai)
print("Số tháng cần gửi: ", thang)
```

```python
import math
def songuyento(n):
    if n<2:
        return False
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return False
    return True
def dem(x, y):
    dem=0
    for i in range(x, y+1):
        if songuyento(i):
            dem+=i
    return dem
print("Số nguyên tố trong đoạn [1,10] là:", dem(1, 10))
print("Số nguyên tố trong đoạn [1,1000] là:", dem(1, 1000))
print("Số nguyên tố trong đoạn [1001,100000] là:", dem(1001,100000))
```

```python
import math
def lasonguyento(x):
    if x < 2:
        return False
    for i in range(2, int(math.sqrt(x)) + 1):
        if x % i == 0:
            return False
    return True
def sothun(n):
    dem = 0
    k = 2
    while True:
        if lasonguyento(k):
            dem += 1
            if dem == n:
                return k
        k += 1
x = int(input("Nhập số: "))
if lasonguyento(x):
    print("Số vừa nhập là:", x, "và là số nguyên tố")
else:
    print("Số vừa nhập là:", x, "và không phải số nguyên tố")
print("Số nguyên tố thứ 1000 là:", sothun(1000))
print("Số nguyên tố thứ 10000 là:", sothun(10000))
```
