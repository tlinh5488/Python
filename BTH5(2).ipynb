Bài 3: Viết chương trình nhập một danh sách sinh viên, mỗi sinh viên gồm: họ tên, năm sinh, đtb. Thực hiện các thao tác sau trên danh sách:
In danh sách sinh viên lên màn hình
Đếm số sinh viên có dtb>=5
Nhập họ tên sinh viên cần tìm x. Tìm sinh viên x trong danh sách.
Nhập họ tên sinh viên cần tìm x. Xóa sinh viên x trong danh sách.
Nhập 1 sinh viên s, và vị trí k, thêm sinh viên s vào danh sách tại vị trí k.



```python
import math

def nhapSV():
    ht = input("Nhập họ và tên: ")
    ns = int(input("Nhập năm sinh: "))
    dtb = float(input("Nhập dtb: "))
    return (ht, ns, dtb)

def nhapDSSV(n):
    ds = []
    for i in range(n):
        s = nhapSV()
        ds.append(s)
    return ds

def inDSSV(ds):
    for s in ds:
        print(s)

def timSV(ds, ht):
    for s in ds:
        if s[0].lower() == ht.lower():
            return s
    return None

def demSV(ds):
    dem = 0
    for s in ds:
        if s[2] >= 5:
            dem += 1
    return dem

def xoaSV(ds, ht):
    ds_moi = []
    for s in ds:
        if s[0].lower() != ht.lower():
            ds_moi.append(s)
    return ds_moi

def themSV(ds, s, k):
    if 0 <= k <= len(ds):
        ds.insert(k, s)
    else:
        print("Vị trí không hợp lệ, thêm vào cuối.")
        ds.append(s)
    return ds


# ===== CHƯƠNG TRÌNH CHÍNH =====

n = int(input("Số sinh viên: "))
ds = nhapDSSV(n)

inDSSV(ds)

print("\nSố lượng sinh viên có đtb >= 5 là:", demSV(ds))

x = input("Nhập họ và tên cần tìm: ")
s = timSV(ds, x)
if s == None:
    print("Không tìm thấy!!!")
else:
    print(s)

x = input("\nNhập họ tên cần xóa: ")
ds = xoaSV(ds, x)
print("Danh sách sau khi xóa:")
inDSSV(ds)

print("\nNhập sinh viên cần thêm:")
s = nhapSV()
k = int(input("Nhập vị trí k: "))
ds = themSV(ds, s, k)

print("Danh sách sau khi thêm:")
inDSSV(ds)
```

Bài 4: Viết chương trình nhập vào 2 tập hợp A và B các số nguyên dương. Tìm giao, hợp, hiệu của 2 tập A và B rồi in lên màn hình.
Hướng dẫn: để nhập 1 tập hợp n phần tử thì xuất phát từ tập rỗng, nhập từng phần tử rồi thêm vào tập. Tìm giao, hợp, hiệu dùng các hàm của tập hợp.


```python
def nhapTap(n, ten):
    s = set()
    print(f"\nNhập tập {ten}:")
    for i in range(n):
        x = int(input(f"Phần tử thứ {i+1}: "))
        if x > 0:
            s.add(x)
        else:
            print("Chỉ nhập số nguyên dương!")
    return s

n = int(input("Nhập số phần tử tập A: "))
A = nhapTap(n, "A")

m = int(input("\nNhập số phần tử tập B: "))
B = nhapTap(m, "B")
giao = A & B
hop = A | B
hieu_AB = A - B
hieu_BA = B - A

print("\nTập A:", A)
print("Tập B:", B)

print("\nGiao A ∩ B:", giao)
print("Hợp A ∪ B:", hop)
print("Hiệu A - B:", hieu_AB)
print("Hiệu B - A:", hieu_BA)
```

Bài 5: Viết chương trình nhập vào từ điển Anh-Việt. Sau đó thực hiện:
In từ điển lên màn hình theo dạng: tiếng Anh: tiếng Việt.
Nhập 1 từ tiếng Anh, cho biết nghĩa tiếng Việt của từ tiếng Anh đó nếu có.
Nhập 1 từ tiếng Anh và nghĩa tiếng Việt, thêm từ đó vào từ điển.


```python
#Tu dien Anh - Viet
def nhapTD(n):
    td = {}
    for i in range(n):
        ta = input("Từ tiếng Anh ")
        tv = input("Nghĩa tiếng Việt ")
        td.update({ta:tv})
    return td
def inTD(td):
    for t in td:
        print(t,":",td[t])

def timTu(td, ta):
    return td.get(ta)

def themTu(td, ta, tv):
    if ta in td:
        print("Thêm từ thành công, hãy nhập thêm nghĩa!")
    td[ta] = tv
    return td

tdAV = nhapTD(3)
inTD(tdAV)

ta = input("\nNhập từ tiếng Anh cần tìm: ")
tv = timTu(tdAV, ta)

if tv == None:
    print("Không tìm thấy từ trong từ điển")
else:
    print(ta, ":", tv)

print("\nNhập từ cần thêm: ")
ta = input("Tiếng Anh: ")
tv = input("Nghĩa tiếng Việt: ")

tdAV = themTu(tdAV, ta, tv)

print("\nTừ điển sau khi thêm:")
inTD(tdAV)
```

Bài 6. XỬ LÝ DANH SÁCH ĐIỂM SINH VIÊN
Cho danh sách:
students = [ ("An", 8.5),
    ("Binh", 6.0),
    ("Cuong", 9.0),
    ("Dung", 4.5),
    ("Hanh", 7.0)
]
Yêu cầu
Tính điểm trung bình
Tìm sinh viên có điểm cao nhất
Lọc sinh viên đạt ≥ 7
Tạo dictionary {tên: điểm}
Chuẩn hóa điểm về thang 0–1
Phân loại học lực


```python
students = [
    ("An", 8.5),
     ("Binh", 6.0),
    ("Cuong", 9.0),
     ("Dung", 4.5),
    ("Hanh", 7.0)
]

dtb = sum(s[1] for s in students) / len(students)
max_sv = max(students, key=lambda x: x[1])
dat = [s for s in students if s[1] >= 7]

d = {s[0]: s[1] for s in students}

ten_set = {s[0] for s in students}

chuanhoa = [(s[0], s[1]/10) for s in students]

def xl(d):
    return "Giỏi" if d>=8 else "Khá" if d>=6.5 else "TB" if d>=5 else "Yếu"

xeploai = [(s[0], xl(s[1])) for s in students]

print("DTB:", dtb)
print("Cao điểm nhất:", max_sv)
print("Đạt:", dat)
print("Dictionary:", d)
print("Set:", ten_set)
print("Chuẩn hóa:", chuanhoa)
print("Xếp loại:", xeploai)
```
