```python
import pandas as pd

hoten = ['Nguyễn Thị Lệ', 'Lê Văn Tài', 'Hồ Hữu Hoàn', 'Lý Thị Lựu', 'Trần Trọng Trí', 'Nguyễn Thị Lệ', 'Phè Văn Phởn']
ngaysinh = ['01/01/45', '21/10/76', '03/03/90', '30/04/01', '01/09/63', '01/01/45', '22/09/04']
gioitinh = ['Nữ', 'Nam','Nam','Nữ', 'Nam','Nữ', 'Nam']
td_CN = [1, 1, 0, 1, 1, 1, 1]
td_ThS = [1, 2, 0, 1, 0, 1, 1]
td_TS = [0, 1, 0, 0, 0, 0, 1]
sodt = ["0915030909", "0902524444", "0232321234", "0170239311", "0904031234", "0915030909", "0907234445"]
dulieu1 = list(zip(hoten, ngaysinh, gioitinh, td_CN, td_ThS, td_TS, sodt))
df = pd.DataFrame(dulieu1, columns=['hoten', 'ngaysinh', 'gioitinh', 'td_CN', 'td_ThS', 'td_TS', 'sodt'])
print(df)
df.to_csv('danhsach.csv', index=False)
```

```python
df_gt = pd.get_dummies(df['gioitinh'])
df_new = df.join(df_gt)
df_new = pd.concat([df, df_gt], axis=1)
# Xây dựng hàm chuyển đổi giá trị
def score_to_numeric(x):
    if x=='Nam':
        return 1
    if x=='Nữ':
        return 0
# Gọi hàm chuyển đổi giá trị cho trường Giới tính
df_new.gioitinh = df_new.gioitinh.apply(score_to_numeric)
df_new.duplicated()
df_trung = df_new.drop_duplicates()
#Tạo tập tin dữ liệu data01.csv
df_trung.to_csv('data01.csv',index=False,header=True)
print(df_trung)
```

```python
location2 = "data01.csv"
df = pd.read_csv(location2)
#Xóa cột Nam Nữ
df = df.drop('Nam', axis=1)
df = df.drop('Nữ', axis=1)

#Thêm cột Hưu trí với giá trị với dãy giá trị cho trước
d = ([1,0,0,1,0,1])
s = pd.Series(d, index= df.index)
df['Huutri'] = s

#Thêm cột hệ số lương với giá trị cho trước
d = ([6.4,3.99,2.97,5.4,4.2,8.2])
s = pd.Series(d, index= df.index)
df['hsl'] = s

#Thêm cột lương cơ bản với giá trị bằng Hệ số lương*1350000
l = df['hsl']*1350000
s = pd.Series(l, index= df.index)
df['luongcb'] = s

#Đổi tên cột sang chữ thường, ký tự đầu là ký tự hoa
df.columns = map(str.lower, df.columns)

# Xây dựng hàm chuyển đổi giá trị
def score_to_num(x):
    if x==0:
        return 0
    if x>=1:
        return 1
print (df)
df.to_csv('data02.csv',index=False,header=True)
```
