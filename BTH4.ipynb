```python
n=int(input("Nhập số sinh viên:"))
diem=[]
for i in range(n):
    x=float(input(f"Nhập điểm sinh viên {i+1}:"))
    diem.append(x)
maxdiem=max(diem)
mindiem=min(diem)
tb=sum(diem)/n
print("Điểm cao nhất:", maxdiem)
print("Điểm thấp nhất:", mindiem)
print("Điểm trung bình:", tb)
```

```python
def sochan(a):
    tong=0
    dem=0
    for x in a:
        if x%2==0:
            tong += x
            dem += 1
    if dem == 0:
        return None
    return tong/dem
n= int(input("Nhập số phần tử trong mảng:"))
a=[]
for i in range(n):
    x=int(input(f"Nhập phần tử {i+1}:"))
    a.append(x)
kq= sochan(a)
if kq==None:
    print("Không có số chẵn trong mảng")
else:
    print("Trung bình các số chẵn:", kq)
```

```python
import math
def songuyento(x):
    if x < 2:
        return False
    for i in range(2, int(math.sqrt(x)) + 1):
        if x % i == 0:
            return False
    return True
n= int(input("Nhập số phần tử:"))
a=[]
for i in range(n):
    x=int(input(f"Nhập phần tử {i+1}:"))
    a.append(x)
m=[]
for x in a:
    if songuyento(x):
        m.append(x)
dem=len(m)
print("Số lượng số nguyên tố:", dem)
if dem>0:
    print("Số nguyên tố lớn nhất: ", max(m))
else:
    print("Không có số nguyên tố trong mảng")
print("Mảng các số nguyên tố:", m)
```
