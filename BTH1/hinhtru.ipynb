```python
import math
r=float(input("Nhập bán kính hình trụ:"))
h=float(input("Nhập chiều cao hình trụ:"))
sXQ=2*math.pi*r*h
sTP=2*math.pi*r*(r+h)
v=math.pi*math.pow(r,2)*h
print("Diện tích xung quanh cuả hình trụ:"+str(sXQ))
print("Diện tích toàn phần:"+str(sTP))
print("Thể tích hình trụ:"+str(v))
```
