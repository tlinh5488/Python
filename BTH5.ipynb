```python
def sinhVien():
    ten= input("Nhập họ tên: ")
    msv=input("Nhập MSV: ")
    dcc=float(input("Nhập điểm chuyên cần: "))
    dgk=float(input("Nhập điểm giữa kì: "))
    dck=float(input("Nhập điểm cuối kì: "))
    return (ten,msv,dcc,dgk,dck)
sv=sinhVien()
print(sv)
```
